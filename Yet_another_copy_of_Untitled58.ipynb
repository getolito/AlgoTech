```python
# Identification du client ayant le ratio le plus eleve
df['Ratio_Credit_revenu'] = df['Credit']/df['Revenu'].replace(0,pd.NA)
client_max_ratio=df.loc[df['Ratio_Credit_revenu'].idxmax()]
print("client ayant le ratio le plus eleve:")
print(client_max_ratio)
```

```python
import numpy as np
```

```python
# segmentation
# Definissons les seuils
seuil_bas = 2000
seuil_haut = 4000
```

```python
# Conditions
conditions = [
    df['Revenu']<seuil_bas,
    (df['Revenu']>=seuil_bas)&(df['Revenu']<seuil_haut),
    df['Revenu']>=seuil_haut
]
```

```python
# Choix
choix = ['Revenu faible', 'Revenu moyen','Revenu eleve']
```

```python
# Creation de la nouvelle variable
df['categorie_revenu'] = np.select(conditions,choix, default='')
```

```python
# affichage de la repartition par categorie
print(df['categorie_revenu'].value_counts())
```

```python
# credit moyen par categorie de revenu
credit_moyen_par_categorie = df.groupby('categorie_revenu')['Credit'].mean()
print("Credit moyen par categorie:")
print(credit_moyen_par_categorie)
```

```python
# taux de defaut par categorie
taux_defaut = df.groupby('categorie_revenu')['Defaut'].mean()*100
print("Taux de defaut par categorie de revenu(%):")
print(taux_defaut.round(2))
```

```python
# extraction des clients ayant un credit superieur a 15000
clients_credit_eleve = df[df['Credit']>15000]
print("Clients avec un credit eleve:")
print(clients_credit_eleve)
```

```python
# determination de leur age moyen
age_moyen_clients_eleves=clients_credit_eleve[['Age']]
print("Age moyen des clients avec un credit eleve:")
print(age_moyen_clients_eleves.mean())
```

```python
# determination de leur revenu moyen
revenu_moyen_clients_eleves=clients_credit_eleve[['Revenu']]
print("Revenu moyen des clients avec un credit eleve:")
print(revenu_moyen_clients_eleves.mean())
```

```python
# Calcul de leur proportion de defaut
proportion_defaut_clients_eleves=clients_credit_eleve[['Defaut']]
print("Proportion de defaut des clients avec un credit eleve:")
print(proportion_defaut_clients_eleves.mean())
```

```python
# Encodage
# creer une variable Risque, Faible si defaut = 0 , Eleve si Defaut = 1
df['Risque']=df['Defaut'].map({0:'Faible', 1:'Eleve'})
print(df[['Defaut','Risque']].head(10))
```

```python
# Verification de la transformation
print(df['Risque'].value_counts())
```

```python
# Affichage de la frequence de chaque categorie
print(df['Risque'].value_counts(normalize=True*100))
```

```python
# Analyse des extremes
# Identification des 3 clients ayant les credits les plus eleves
clients_credit_eleves=df.nlargest(3,'Credit')
print("Les 3 clients avec les credits les plus eleves:")
print(clients_credit_eleves)
```

```python
# Ayant les credits les plus faibles
clients_credit_eleves=df.nsmallest(3,'Credit')
print("Les 3 clients avec les credits les plus eleves:")
print(clients_credit_eleves)
```

```python
import pandas as pd
# Rechargeons le jeux de donnees initial en vue d'eviter des erreurs dans les noms
df = pd.read_excel('base_100.xlsx')
print(df)
# Creons un nouveau fichie Excel contenant les donnees nettoyees avec les nouvelles variables : Ratio_Credit_Revenu, Categorie_Revenu,Risque
df.to_excel('dataset_nettoye.xlsx', index=False, sheet_name='base_100_analyse')
print("Apercu des donnees nettoyees")
print(df.head())
```

```python
# creation de Ratio_Credit_revenu
# evitons la division par zero
df['Ratio_Credit_revenu']=df['Credit']/df['Revenu'].replace(0,np.nan)
# creation de Categorie_Revenu
seuil_bas=2000 # en dessous:"Revenu faible"
seuil_haut=4000 # au-dessus : :"Revenu eleve"
# creation des intervalles
bins=[-float('inf'),seuil_bas,seuil_haut,float('inf')]
labels=['Revenu faible','Revenu moyen','Revenu eleve']
df['Categorie_Revenu']=pd.cut(df['Revenu'],bins=bins,labels=labels,right=False)
#creation de risque
df['Risque']=df['Defaut'].map({0:'Faible',1:'Eleve'})
# Rapide verification
print("Apercu des nouvelles colonnes")
print(df[['Credit','Revenu','Defaut','Ratio_Credit_revenu','Categorie_Revenu','Risque']].head(10))
```

```python
print("\nRepartition par categorie de revenu:")
print(df['Categorie_Revenu'].value_counts())
print("\nRepartition du risque:")
print(df['Risque'].value_counts())
```

```python
# Jeux de donnees nettoyees avec les nouvelles variables
df.to_excel('dataset_nettoye.xlsx',index=False,sheet_name='base_100_analyse.xlsx')
```
