Forte variabilite des montants de credit

```python
# Comparaison du revenu moyen entre les deux groupes(Defaut = 0 vs Defaut = 1)
moyennes = df.groupby('Defaut')['Revenu'].mean()
print("Revenu moyen par groupe de defaut:")
print(moyennes)
```

```python
# Comparaison du credit moyen entre les deux groupes(Defaut = 0 vs Defaut =1)
moyennes = df.groupby('Defaut')['Credit'].mean()
print("Credit moyen par groupe de defaut")
print(moyennes)
```

On constate que les clients bon payeurs i.e qui remboursent leurs credits ont des revenus moyens inferieurs aux clients qui ne remboursent pas leurs credits.
